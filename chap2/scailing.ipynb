```python
from sklearn import datasets, linear_model
import timeit
```

원본데이터가 442 * 10, 3번째 열이 BMI 데이터

이때 diabetes_X[:, 2] 만을 진행하면 [442,1]인 데이터가 아니라,

파이썬에서 자동으로 [442] 꼴의 데이터로 변환, 따라서 None을 넣어 한 차원 추가하여 
세로로 세워진 열 형태로 유지

이때 None은 np.newaxis와 똑같은 역할

```python
# returen_X_y=True : data(features), target(label)을 각각 반환
diabetes_X, diabetes_y = datasets.load_diabetes(return_X_y=True)

# 2 : 데이터 넷의 3번째 특성만 사용,
# None : 1 차원배열을 2차원 행렬 구조로 바꿈, scikit-learn은 2차원 행렬 구조를 요구
# : : 모든 행 선택 
raw = diabetes_X[:, None, 2]
max_raw = max(raw)
min_raw = min(raw)

scaled = 2 * (raw - max_raw - min_raw) / (max_raw - min_raw)

def train_raw():
    linear_model.LinearRegression().fit(raw, diabetes_y)

def train_scaled():
    linear_model.LinearRegression().fit(scaled, diabetes_y)

raw_time = timeit.timeit(train_raw, number=1000)
scaled_time = timeit.timeit(train_scaled, number=1000)

print(f"Raw data training time: {raw_time:.4f} seconds")
print(f"Scaled data training time: {scaled_time:.4f} seconds")
```

입력 데이터만을 스케일링 했는데 모델에서 스케일링 한 쪽의 속도가 조금 향상된 것을 확인할 수 있음
